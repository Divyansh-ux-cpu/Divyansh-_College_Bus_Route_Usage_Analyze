## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_17_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
# TODO: Display the shape and column names
# TODO: Inspect data types and missing values
# TODO: Count duplicate row

# Display number of rows and columns
print("Dataset Shape:", df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display information about dataset
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# Display first five records
print("\nFirst Five Records:")
display(df.head())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python

# Select numeric columns
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numeric Columns:")
print(numeric_columns.tolist())

# Convert data into NumPy array
data_array = df[numeric_columns].to_numpy()

print("\nNumPy Array:")
print(data_array)

# Calculate average
print("\nAverage Values:")
print(np.nanmean(data_array, axis=0))

# Calculate maximum
print("\nMaximum Values:")
print(np.nanmax(data_array, axis=0))

# Calculate minimum
print("\nMinimum Values:")
print(np.nanmin(data_array, axis=0))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# 5. Pandas Analysis

# Calculate passengers per KM
df["Passengers_Per_KM"] = (
    df["Daily_Passengers"] / df["Distance_KM"]
)

# Calculate average passengers
average_passengers = df["Daily_Passengers"].mean()

# Filter high usage records
high_usage = df[
    df["Daily_Passengers"] > average_passengers
]

# Sort routes by passengers per KM
top_routes = df.sort_values(
    by="Passengers_Per_KM",
    ascending=False
)

# Display results
print("Average Passengers:", average_passengers)

print("\nHigh Usage Records:")
display(high_usage)

print("\nTop 5 Records:")
display(top_routes.head(5))
```

## 6. Your individual project challenge

```python
# 6. Individual Project Challenge

# Find total passengers
total_passengers = df["Daily_Passengers"].sum()

# Find average passengers
average_passengers = df["Daily_Passengers"].mean()

# Find maximum passengers
maximum_passengers = df["Daily_Passengers"].max()

# Find minimum passengers
minimum_passengers = df["Daily_Passengers"].min()

# Display results
print("Total Passengers:", total_passengers)
print("Average Passengers:", average_passengers)
print("Maximum Passengers:", maximum_passengers)
print("Minimum Passengers:", minimum_passengers)

# Display top 5 busiest records
busiest_records = df.sort_values(
    by="Daily_Passengers",
    ascending=False
)

display(busiest_records.head(5))
```
